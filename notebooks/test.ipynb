```python
import json
from pathlib import Path
import sys
import pandas as pd
parentDir = Path.cwd().parent
sys.path.insert(0, str(parentDir))
```

```python
path = parentDir / "data/eeg_stream_walking.jsonl"

# Load each JSON line into a list of dicts
records = []
with open(path, "r") as f:
    for line in f:
        if line.strip():
            records.append(json.loads(line))

# Convert to DataFrame
df = pd.DataFrame(records)

print(df.head())
print(df.shape)
```
